# Lab 5 — Visualization & Mini-EDA

Visualize the Wine Quality dataset using histograms, bar charts, scatter plots, and box plots, then summarize key patterns using grouped statistics.

```python
# ============================================================
# 05 - Visualization & Mini-EDA
# Wine Quality Dataset
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../week2-data-processing/data/WineQT.csv")

# ============================================================
# 1. Quality Distribution
# ============================================================

quality_counts = df["quality"].value_counts().sort_index()

print("Quality Distribution:")
print(quality_counts)

plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="quality")
plt.title("Wine Quality Distribution")
plt.xlabel("Quality")
plt.ylabel("Count")
plt.show()

# ============================================================
# 2. Feature Distribution
# ============================================================

plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="alcohol", bins=20, kde=True)
plt.title("Alcohol Distribution")
plt.xlabel("Alcohol")
plt.ylabel("Frequency")
plt.show()

# ============================================================
# 3. Relationship Between Features
# ============================================================

plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df,
    x="alcohol",
    y="quality"
)
plt.title("Alcohol vs Wine Quality")
plt.xlabel("Alcohol")
plt.ylabel("Quality")
plt.show()

# ============================================================
# 4. Box Plot
# ============================================================

plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="quality", y="alcohol")
plt.title("Alcohol Content by Wine Quality")
plt.xlabel("Quality")
plt.ylabel("Alcohol")
plt.show()

# ============================================================
# 5. Grouped Statistics
# ============================================================

print("\nMean Alcohol by Quality:")
print(df.groupby("quality")["alcohol"].mean())

print("\nMedian Alcohol by Quality:")
print(df.groupby("quality")["alcohol"].median())

print("\nNumber of Wines by Quality:")
print(df.groupby("quality")["alcohol"].count())
```

The scatter plot shows a positive relationship between alcohol content and wine quality, although the relationship is not perfectly linear.

## Conclusion

The visualizations show the distribution of wine quality and alcohol content, the relationship between alcohol and quality, and the spread of alcohol content across quality groups. Higher-quality wines generally have higher alcohol levels, while most wines fall within quality levels 5 and 6.
