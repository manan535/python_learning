```python
# ============================================================
# 01 - Pandas Fundamentals
# Wine Quality Dataset
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../week2-data-processing/data/WineQT.csv")

# ============================================================
# 1. Dataset Overview
# ============================================================

print("Dataset Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nData Types and Structure:")
df.info()

print("\nStatistical Summary:")
display(df.describe().T)

# ============================================================
# 2. Column Selection
# ============================================================

print("\nSelected Columns:")
display(df[["alcohol", "pH", "quality"]].head())

print("\nUsing loc:")
display(df.loc[0:4, ["alcohol", "quality"]])

print("\nUsing iloc:")
display(df.iloc[0:5, 0:3])

# ============================================================
# 3. Filtering
# ============================================================

print("\nQuality >= 7:")
display(df[df["quality"] >= 7])

print("\nQuality >= 7 and Alcohol > 10:")
display(
    df[
        (df["quality"] >= 7) &
        (df["alcohol"] > 10)
    ]
)

# ============================================================
# 4. GroupBy
# ============================================================

print("\nAlcohol Statistics by Quality:")
display(
    df.groupby("quality")["alcohol"]
    .agg(["mean", "count"])
)

# ============================================================
# 5. Data Quality
# ============================================================

print("\nMissing Values:")
display(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

# ============================================================
# 6. Merge DataFrames
# ============================================================

df1 = df[["Id", "quality"]]
df2 = df[["Id", "alcohol"]]

merged = pd.merge(df1, df2, on="Id")

print("\nMerged DataFrame:")
display(merged.head())

print("\nMerged Shape:")
print(merged.shape)

# ============================================================
# 7. Quality Distribution
# ============================================================

quality_counts = df["quality"].value_counts().sort_index()

print("\nQuality Distribution:")
print(quality_counts)

plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="quality")
plt.title("Wine Quality Distribution")
plt.xlabel("Quality")
plt.ylabel("Count")
plt.show()

# ============================================================
# 8. Alcohol Distribution
# ============================================================

plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="alcohol", bins=20, kde=True)
plt.title("Alcohol Distribution")
plt.xlabel("Alcohol")
plt.ylabel("Frequency")
plt.show()
```
