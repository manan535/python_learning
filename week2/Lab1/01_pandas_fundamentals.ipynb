```python
# ============================================================
# 01 - Pandas Fundamentals
# Wine Quality Dataset
# ============================================================

import pandas as pd

df = pd.read_csv("../week2-data-processing/data/WineQT.csv")

# ============================================================
# 1. Load and Inspect DataFrame
# ============================================================

print("Dataset Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nData Types and Structure:")
df.info()

print("\nStatistical Summary:")
display(df.describe().T)

# ============================================================
# 2. Select and Filter Rows and Columns
# ============================================================

print("\nSelected Columns:")
display(df[["alcohol", "pH", "quality"]].head())

print("\nUsing loc:")
display(df.loc[0:4, ["alcohol", "quality"]])

print("\nUsing iloc:")
display(df.iloc[0:5, 0:3])

print("\nQuality >= 7:")
display(df[df["quality"] >= 7])

print("\nQuality >= 7 and Alcohol > 10:")
display(
    df[
        (df["quality"] >= 7) &
        (df["alcohol"] > 10)
    ]
)

# ============================================================
# 3. Create a Derived Column
# ============================================================

df["alcohol_high"] = df["alcohol"] > 10

print("\nDerived Column:")
display(df[["alcohol", "alcohol_high"]].head())

# ============================================================
# 4. Group and Aggregate
# ============================================================

print("\nAlcohol Statistics by Quality:")
display(
    df.groupby("quality")["alcohol"]
    .agg(["mean", "count"])
)

# ============================================================
# 5. Merge Two DataFrames
# ============================================================

df1 = df[["Id", "quality"]]
df2 = df[["Id", "alcohol"]]

merged = pd.merge(df1, df2, on="Id")

print("\nMerged DataFrame:")
display(merged.head())

merged = pd.merge(df1, df2, on="Id")

print("Merged shape:", merged.shape)
print("Merged row count makes sense because Id is unique in both DataFrames.")
```
