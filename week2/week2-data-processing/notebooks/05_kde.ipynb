```python
# ============================================================
# 05 - Kernel Density Estimation (KDE)
# Wine Quality Dataset
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/WineQT.csv")

# ============================================================
# 1. KDE of Alcohol
# ============================================================

plt.figure(figsize=(8, 5))

sns.kdeplot(
    data=df,
    x="alcohol"
)

plt.title("KDE of Alcohol")
plt.xlabel("Alcohol")
plt.ylabel("Density")
plt.show()

# ============================================================
# 2. Histogram + KDE
# ============================================================

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="alcohol",
    bins=20,
    stat="density",
    kde=True
)

plt.title("Alcohol Distribution with KDE")
plt.xlabel("Alcohol")
plt.ylabel("Density")
plt.show()

# ============================================================
# 3. KDE with Different Bandwidths
# ============================================================

plt.figure(figsize=(10, 6))

for bandwidth in [0.5, 1, 2]:
    sns.kdeplot(
        data=df,
        x="alcohol",
        bw_adjust=bandwidth,
        label=f"bw_adjust = {bandwidth}"
    )

plt.title("Effect of Bandwidth on KDE")
plt.xlabel("Alcohol")
plt.ylabel("Density")
plt.legend()
plt.show()

# ============================================================
# 4. KDE for Multiple Features
# ============================================================

features = [
    "fixed acidity",
    "volatile acidity",
    "density",
    "pH",
    "sulphates",
    "alcohol"
]

fig, axes = plt.subplots(2, 3, figsize=(15, 8))

for ax, feature in zip(axes.flat, features):
    sns.kdeplot(
        data=df,
        x=feature,
        ax=ax
    )
    ax.set_title(feature)

plt.tight_layout()
plt.show()

# ============================================================
# 5. KDE by Wine Quality
# ============================================================

plt.figure(figsize=(10, 6))

sns.kdeplot(
    data=df,
    x="alcohol",
    hue="quality",
    common_norm=False
)

plt.title("Alcohol Density by Wine Quality")
plt.xlabel("Alcohol")
plt.ylabel("Density")
plt.show()

# ============================================================
# 6. 2D KDE
# ============================================================

plt.figure(figsize=(9, 6))

sns.kdeplot(
    data=df,
    x="alcohol",
    y="density",
    fill=True
)

plt.title("2D KDE: Alcohol vs Density")
plt.xlabel("Alcohol")
plt.ylabel("Density")
plt.show()

# ============================================================
# 7. KDE Interpretation
# ============================================================

print("""
KDE represents the estimated probability density of a variable.

Important observations:
- Higher peaks represent regions where observations are more concentrated.
- Lower peaks represent regions with fewer observations.
- Smaller bandwidths produce a more detailed and potentially noisier curve.
- Larger bandwidths produce a smoother curve but can hide smaller patterns.
""")
```
