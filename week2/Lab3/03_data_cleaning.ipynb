# Lab 3 — Data Cleaning

Clean a messy dataset by profiling missing values, fixing data types, handling duplicates and outliers, and validating the final DataFrame.

```python
import pandas as pd

df = pd.read_csv("data.csv")

print("Original Shape:", df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nMissing Value Percentage:")
print((df.isnull().mean() * 100).round(2))

df["age"] = pd.to_numeric(df["age"], errors="coerce")
df["salary"] = pd.to_numeric(df["salary"], errors="coerce")
df["join_date"] = pd.to_datetime(df["join_date"], errors="coerce")

df["age"] = df["age"].fillna(df["age"].median())
df["salary"] = df["salary"].fillna(df["salary"].median())
df["join_date"] = df["join_date"].fillna(df["join_date"].mode()[0])

duplicate_count = df.duplicated().sum()
df = df.drop_duplicates()

q1 = df["age"].quantile(0.25)
q3 = df["age"].quantile(0.75)
iqr = q3 - q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

outlier_count = ((df["age"] < lower) | (df["age"] > upper)).sum()

df = df[df["age"].between(18, 100)]

print("\nCleaning Decisions:")
print("Age: median imputation preserves the typical age without removing rows.")
print("Salary: median imputation reduces the effect of extreme salaries.")
print("Join date: mode imputation replaces the invalid date with the most common valid date.")
print("Duplicates: duplicate rows were removed to avoid counting the same record twice.")
print("Age outliers: unrealistic ages were removed because they fall outside a sensible human age range.")

print("\nDuplicates Removed:", duplicate_count)
print("Age Outliers Detected:", outlier_count)

print("\nData Types:")
print(df.dtypes)

print("\nRemaining Missing Values:")
print(df.isnull().sum())

print("\nCleaned Data:")
display(df)

print("\nFinal Shape:", df.shape)

print("\nFinal Summary:")
display(df.describe())
```

## Conclusion

The dataset was cleaned by imputing appropriate missing values, correcting data types, removing duplicate records and unrealistic ages, and validating the final DataFrame for missing values and sensible ranges.
